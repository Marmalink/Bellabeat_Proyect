```python
# Creo dataframe calories
calories = pd.read_csv('/home/Escritorio/ProyectoBellaBeat/Fitabase Data 4.12.16-5.12.16/hourlyCalories_merged.csv')
print(calories)
```

```python
calories.isnull().sum()
# No hay valores vacios
```

```python
# Cambio el formato de la columna ActivityHour
calories["ActivityHour"] = pd.to_datetime(calories["ActivityHour"])
```

```python
# Creo la columna ActivityHourFormatted con los datos de la columna ActivityHour
calories["ActivityHourFormatted"] = calories["ActivityHour"].dt.strftime("%d/%m/%Y %H:%M:%S")
print(calories)
```

```python
# Elimino la columna ActivityHour
calories.drop(['ActivityHour'], axis=1, inplace=True)
```

```python
# Creo un nuevo dataframe para separar los datos de fecha y horario
formatted=calories["ActivityHourFormatted"].str.split(expand=True)
formatted.columns=['ActivityDate', 'ActivityHour']
calories=pd.concat([calories,formatted], axis=1)
```

```python
# Elimino la columna ActivityHourFormatted ya que se reemplazará por las nuevas columnas
calories.drop(['ActivityHourFormatted'], axis=1, inplace=True)
```

```python
print(calories)
```
