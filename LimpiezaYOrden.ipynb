## Calories data

```python
calories = pd.read_csv('Fitabase Data 4.12.16-5.12.16/hourlyCalories_merged.csv')
print(calories)
```

```python
calories.isnull().sum()
# No hay valores vacios
```

```python
# Cambio el formato de la columna ActivityHour
calories["ActivityHour"] = pd.to_datetime(calories["ActivityHour"])
```

```python
# Creo la columna ActivityHourFormatted con los datos de la columna ActivityHour
calories["ActivityHourFormatted"] = calories["ActivityHour"].dt.strftime("%d/%m/%Y %H:%M:%S")
```

```python
# Creo una nueva columna para separar los datos de fecha y horario
formatted=calories["ActivityHourFormatted"].str.split(expand=True)
formatted.columns=['Activity_Date', 'Activity_Hour']
calories=pd.concat([calories,formatted], axis=1)
```

```python
# Elimino las columna ActivityHourFormatted y ActivityHour ya que se reemplazará por las nuevas columnas
calories.drop(['ActivityHourFormatted', 'ActivityHour'], axis=1, inplace=True)
```

```python
calories['Codigo_Id'] = calories.groupby('Id').ngroup()
```

```python
print(calories)
```
